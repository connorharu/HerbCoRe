## First step

In `first_step.ipynb`, we will collect the speciesLink records that will be used in our database.

```python
from library import *

specieslink, db_config = configure()
```

```python
def selection(filters_input, table):
    filters = {}
    for item in filters_input.split():
        if '=' not in item: 
            print(f"badly formatted filter: {item} - use key=value")
            break
        else:
            key, value = item.split('=', 1)
            filters[key.strip()] = value.strip()

    records = specieslink.search_records(filters=filters)

    if records:
        print(f"\n{len(records)} records found, inserting in '{table}'...")
        specieslink.insert_into_mysql(records, db_config, table=table)
        print("records inserted successfully.")
    else:
        print("no records found with the provided filters.")
```

```python
filters_input = input("provide the filters (key=value format, separate filters with a blank space): ").strip()
table = "biodiversity_records" # provided in biodiversity_records.sql

print(f"executing specieslink.search_records(filters={filters_input}, table={table})...\n")
selection(filters_input, table)
```
